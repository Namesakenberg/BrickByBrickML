```python
import pandas as pd
import numpy as np
```

```python
flats = pd.read_csv('/content/flats_cleaned.csv')
houses = pd.read_csv('/content/house_cleaned.csv')
```

```python
merged_data = pd.concat([flats,houses],ignore_index=True)     # merging the houses and flats
```

```python
merged_data = merged_data.sample(merged_data.shape[0],ignore_index=True)    #shufflig the merged data
```
