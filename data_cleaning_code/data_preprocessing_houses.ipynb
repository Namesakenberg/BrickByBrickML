```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df.drop(columns=['link','property_id'],inplace=True)
```

```python
df.head()
```

## Society

```python
df.rename(columns = {'rate':'price_per_sqft'},inplace=True)
df.head()
```

```python
import re
df['society'] = df['society'].apply(lambda name:re.sub(r'\d+(\.\d+)?\s?★','',str(name)).strip()).str.lower()
```

```python
df['society'].value_counts().shape
```

```python
df['society'] = df['society'].replace('nan','independent')
```

```python
print(df['society'].value_counts())
```

## Price

```python
print(df['price'].isnull().sum())
```

```python
df = df[df['price']!='Price on Request']
```

```python
def treat_price(x):
  x=str(x).split()
  if len(x) !=2 :
    return None
  num,unit= x
  num = float(num)
  if unit =='Lac':
    return round(num/100,2)
  elif unit =='Crore':
    return round(num,2)
  else:
    return None
  return None
df['price'] = df['price'].apply(treat_price)
```

## Price per sqft

```python
print(df['price_per_sqft'].value_counts())
```

```python
print(df['price_per_sqft'].isnull().sum())
```

```python
df['price_per_sqft'] = df['price_per_sqft'].str.split('/').str.get(0).str.replace('₹','').str.replace(',','').str.strip().astype('float')
```

# Bedrooms , bathrooms and balcony

```python
print(df['bedRoom'].isnull().sum())
print(df['bathroom'].isnull().sum())
print(df['balcony'].isnull().sum())
```

```python
print(df['bedRoom'].value_counts())
print(df['bathroom'].value_counts())
print(df['balcony'].value_counts())
```

```python
df=df[~df['bedRoom'].isnull()]
```

```python
df.shape
```

```python
df.loc[:,'bedRoom']=df['bedRoom'].str.split(' ').str.get(0).astype('int')
df.loc[:,'bathroom']=df['bathroom'].str.split(' ').str.get(0).astype('int')
df.loc[:,'balcony'] = df['balcony'].str.split(' ').str.get(0).str.replace('No','0')
```

```python
print(df['additionalRoom'].value_counts())
```

```python
print(df['additionalRoom'].isnull().sum())
```

```python
df['additionalRoom'] = df['additionalRoom'].fillna('not available')
```

```python
df.loc[:,'additionalRoom']=df['additionalRoom'].str.lower()
```

```python
print(df['noOfFloor'].value_counts())
```

```python
print(df['noOfFloor'].isnull().sum())
```

```python
df.loc[:,'noOfFloor'] = df['noOfFloor'].str.split(' ').str.get(0)
```

```python
df.rename(columns={'noOfFloor':'floorNum'},inplace=True)
```

```python
df.loc[:,'area'] = round((df['price']*10000000 )/df['price_per_sqft'])
```

```python
df.loc[:,'facing'] = df['facing'].fillna('NA')
```

```python
df.insert(loc=1,column='property_type',value='house')
```
