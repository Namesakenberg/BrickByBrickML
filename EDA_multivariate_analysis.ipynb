```python
# calculate the average price of every sector
# sort it numerically
# reperesent in the form of the heatmap

import re

# groupby the sector and the price to get the avg price of each sector

avg_price_by_sector = df.groupby('sector')['price'].mean().reset_index()

# get the sector number
def get_sector_number(sector_name):
  match = re.search(r'\d+',sector_name)
  if match:
    return int(match.group())
  else:
    return float('-inf')
avg_price_by_sector['sector_number'] = avg_price_by_sector['sector'].apply(get_sector_number)

# sort by the sector number
avg_price_by_sector_sorted = avg_price_by_sector.sort_values('sector_number')

# plot the heatmap
plt.figure(figsize=(5,25))
sns.heatmap(avg_price_by_sector_sorted.set_index('sector')[['price']],annot=True , fmt=".2f" ,linewidths=.5)
plt.title('Average price per sector')
plt.xlabel('Average Price')
plt.ylabel('Sector')
plt.show()
```

```python
# calculate the average priceper sqft of every sector
# sort it numerically
# reperesent in the form of the heatmap

import re

# groupby the sector and the price to get the avg price of each sector

avg_price_per_sqft_by_sector = df.groupby('sector')['price_per_sqft'].mean().reset_index()


avg_price_per_sqft_by_sector['sector_number'] = avg_price_per_sqft_by_sector['sector'].apply(get_sector_number)

# sort by the sector number
avg_price_per_sqft_by_sector_sorted = avg_price_per_sqft_by_sector.sort_values('sector_number')

# plot the heatmap
plt.figure(figsize=(5,25))
sns.heatmap(avg_price_per_sqft_by_sector_sorted.set_index('sector')[['price_per_sqft']],annot=True , fmt=".2f" ,linewidths=.5)
plt.title('Average price per sqft per sector')
plt.xlabel('Average Price per sqft')
plt.ylabel('Sector')
plt.show()
```

```python
# avg luxury score per sector

luxury_score = df.groupby('sector')['luxury_score'].mean().reset_index()

luxury_score['sector_number'] = luxury_score['sector'].apply(get_sector_number)

# Sort by sector number
luxury_score_sector = luxury_score.sort_values(by='sector_number')

# Plot the heatmap
plt.figure(figsize=(5, 25))
sns.heatmap(luxury_score_sector.set_index('sector')[['luxury_score']], annot=True, fmt=".2f", linewidths=.5)
plt.title('Sector (Sorted by Sector Number)')
plt.xlabel('Average luxury score')
plt.ylabel('Sector')
plt.show()
```
